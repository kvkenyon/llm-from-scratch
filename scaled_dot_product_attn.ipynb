```python
import torch
from llm.nn import softmax
from einops import rearrange
```

```python
batch_size, seq_len, d_k, d_v = 1, 5, 8, 4
```

```python
Q = torch.randn(batch_size, seq_len, d_k)
K = torch.randn(batch_size, seq_len, d_k)
V = torch.randn(batch_size, seq_len, d_v)
M = torch.randn(seq_len, seq_len)
M = M < 0
# Add leading dimensions of size 1 until counts match
diff = Q.ndim - M.ndim
new_shape = (1,) * diff + M.shape
M = M.reshape(new_shape)  # or x.view(new_shape)

K_T = rearrange(K, '... seq_len d_k -> ... d_k seq_len')
```

```python
pre_softmax_attn = (Q @ K_T) / torch.sqrt(torch.tensor(d_k))
pre_softmax_attn[M == False] += float('-inf')
attn = softmax(pre_softmax_attn, dim=-1)
```
