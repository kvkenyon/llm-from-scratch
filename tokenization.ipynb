```python
from llm.tokenizer import Tokenizer, benchmark_tokenizer
from llm.common import serialize_token_ids
```

```python
dataset = "owt_train.txt"

tokenizer = Tokenizer.from_files("vocab_owt_train.json", "owt_train_merges.txt")
```

```python
tokenizer.encode("bah")
```

```python
tokenizer.decode(_)
```

```python
result = benchmark_tokenizer(tokenizer, dataset)
```

```python
result
```

```python
mib_per_sec = result['mib_per_second']
gib_in_mib = 1e9 / 1000 ** 2
```

```python
hours = (((gib_in_mib / mib_per_sec) / 60) * 12) / 60
print(f"hours to encode owt: {hours}")
```

```python
import numpy as np
```

```python
all_ids = []
with open("data/owt_train.txt") as f:
    for _id in tokenizer.encode_iterable(f):
        all_ids.append(_id)
arr = np.array(all_ids, dtype=np.uint16)

serialize_token_ids("owt.npy", arr)
```
